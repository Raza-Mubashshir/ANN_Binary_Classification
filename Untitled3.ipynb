```python
# ANN for Binary Classification using Back-Propagation
# Dataset: Iris Dataset
# Classes used: Setosa (0) and Versicolor (1)

import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report


# ---------------------------------------------------------
# 1. Load Iris Dataset
# ---------------------------------------------------------

iris = load_iris()

X = iris.data
y = iris.target

# Select only Setosa (0) and Versicolor (1)
mask = y < 2

X = X[mask]
y = y[mask]

# Convert target into column vector
y = y.reshape(-1, 1)

print("Dataset shape:", X.shape)
print("Classes:", np.unique(y))


# ---------------------------------------------------------
# 2. Split Dataset
# ---------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ---------------------------------------------------------
# 3. Feature Scaling
# ---------------------------------------------------------

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)


# ---------------------------------------------------------
# 4. Sigmoid Activation Function
# ---------------------------------------------------------

def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def sigmoid_derivative(x):
    return x * (1 - x)


# ---------------------------------------------------------
# 5. Initialize ANN
# ---------------------------------------------------------

np.random.seed(42)

input_neurons = 4
hidden_neurons = 8
output_neurons = 1

learning_rate = 0.1
epochs = 3000

# Initialize weights
W1 = np.random.randn(input_neurons, hidden_neurons) * 0.1
W2 = np.random.randn(hidden_neurons, output_neurons) * 0.1

# Initialize biases
b1 = np.zeros((1, hidden_neurons))
b2 = np.zeros((1, output_neurons))


# ---------------------------------------------------------
# 6. Training using Back-Propagation
# ---------------------------------------------------------

loss_history = []

for epoch in range(epochs):

    # -------- Forward Propagation --------

    # Input layer -> Hidden layer
    z1 = np.dot(X_train, W1) + b1
    a1 = sigmoid(z1)

    # Hidden layer -> Output layer
    z2 = np.dot(a1, W2) + b2
    y_pred = sigmoid(z2)


    # -------- Calculate Loss --------

    epsilon = 1e-8

    loss = -np.mean(
        y_train * np.log(y_pred + epsilon)
        + (1 - y_train) * np.log(1 - y_pred + epsilon)
    )

    loss_history.append(loss)


    # -------- Back-Propagation --------

    # Output layer error
    error_output = y_pred - y_train

    # Gradients for output layer
    dW2 = np.dot(a1.T, error_output) / len(X_train)
    db2 = np.mean(error_output, axis=0, keepdims=True)

    # Hidden layer error
    error_hidden = np.dot(
        error_output, W2.T
    ) * sigmoid_derivative(a1)

    # Gradients for hidden layer
    dW1 = np.dot(X_train.T, error_hidden) / len(X_train)
    db1 = np.mean(error_hidden, axis=0, keepdims=True)


    # -------- Update Weights --------

    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2

    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1


    # Print loss
    if epoch % 500 == 0:
        print("Epoch:", epoch, "Loss:", loss)


# ---------------------------------------------------------
# 7. Prediction Function
# ---------------------------------------------------------

def predict(X):

    # Forward propagation

    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, W2) + b2
    output = sigmoid(z2)

    # Convert probability into class
    predictions = (output >= 0.5).astype(int)

    return predictions


# ---------------------------------------------------------
# 8. Test the ANN
# ---------------------------------------------------------

y_pred_test = predict(X_test)

accuracy = accuracy_score(y_test, y_pred_test)

print("\n================================")
print("ANN Binary Classification Result")
print("================================")

print("Accuracy:", accuracy * 100, "%")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_test))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_test,
    target_names=["Setosa", "Versicolor"]
))


# ---------------------------------------------------------
# 9. Plot Training Loss
# ---------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(loss_history)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss using Back-Propagation")

plt.grid(True)
plt.show()
```
